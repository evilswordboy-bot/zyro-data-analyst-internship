### 2. Import Libraries
Importing the standard data analysis and visualization stack with conventional aliases.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Visual formatting defaults
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (8, 4.5)
plt.rcParams["font.size"] = 11
print("Libraries successfully imported with visual presets configured.")
```

### 3. Load CSV Dataset
Loading the sales transaction dataset (`sales_data.csv`) and inspecting the first 5 records.

```python
# Load CSV dataset
df = pd.read_csv("sales_data.csv")

# Display first 5 rows
df.head()
```

### 4. Dataset Structure
Examining the dimensional shape, column inventory, and data types of the loaded dataset.

```python
print(f"Dataset Shape (Rows, Columns): {df.shape}")
print("\nColumn Names:")
print(list(df.columns))
print("\nData Types:")
print(df.dtypes)
```

### 5. Missing Values
Validating data integrity by checking for null/missing values across all fields.

```python
missing_counts = df.isnull().sum()
print("Missing Values per Column:")
print(missing_counts)
```

### 6. Summary Statistics
Calculating descriptive numerical statistics (mean, std, min, max, percentiles) and category-level aggregates.

```python
print("Descriptive Statistics:")
display(df.describe())

category_sales = df.groupby('category', as_index=False)['total_sales'].sum()
print("\nTotal Revenue by Category:")
display(category_sales)
```

### 7. Visualization
Generating a clean, readable bar chart visualizing Total Revenue by Product Category.

```python
fig, ax = plt.subplots(figsize=(8, 4.5))

bars = sns.barplot(
    data=category_sales,
    x='category',
    y='total_sales',
    hue='category',
    palette='Blues_d',
    legend=False,
    ax=ax
)

ax.set_title("Total Sales Revenue by Product Category", fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel("Product Category", fontsize=12, labelpad=10)
ax.set_ylabel("Total Sales (USD $)", fontsize=12, labelpad=10)

# Add value labels above bars
for bar in bars.patches:
    height = bar.get_height()
    ax.annotate(f"${height:,.2f}",
                xy=(bar.get_x() + bar.get_width() / 2, height),
                xytext=(0, 6),
                textcoords="offset points",
                ha='center', va='bottom',
                fontweight='semibold')

plt.tight_layout()
os.makedirs("evidence", exist_ok=True)
plt.savefig("evidence/jupyter_analysis.png", dpi=150)
plt.show()
```
